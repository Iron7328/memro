В соревновании в качестве метрики качества использовалось RMSLE:
$$\text{RMSLE}(X, y, a) = \sqrt{\frac{1}{\ell}\sum_{i=1}^{\ell} \big(\log{(y_i + 1)} - \log{(a(x_i) + 1)}\big)^2}$$

**Вопрос (0.1 балл)**: **(?)** Как вы думаете, почему авторы соревнования выбрали именно RMSLE, а не RMSE?

**Ответ:** Потому что RMSLE меньше чувствителен к подобным выбросам.

На семинаре мы рассматривали несколько моделей линейной регрессии в `sklearn`, но каждая из них оптимизировала среднеквадратичную ошибку (MSE), а не RMSLE. Давайте проделаем следующий трюк: будем предсказывать не целевую переменную, а ее *логарифм*. Обозначим $\hat{y}_i = \log{(y_i + 1)}$ — модифицированный таргет, а $\hat{a}(x_i)$ — предсказание модели, которая обучалась на $\hat{y}_i$, то есть логарифм таргета. Чтобы предсказать исходное значение, мы можем просто взять экспоненту от нашего предсказания: $a(x_i) = \exp(\hat{a}(x_i)) - 1$.

**Вопрос (0.3 балла)**: **(?)** Покажите, что оптимизация RMSLE для модели $a$ эквивалентна оптимизации MSE для модели $\hat{a}$.

**Доказательство**: Напишие прямо тут в ячейке, при необходимости использовав синтаксис LaTeX в ячейке

╰( ͡° ͜ʖ ͡° )つ──☆*:・ﾟ

Обозначим исходный таргет $y_i$, модель, предсказывающую исходное значение, — $a(x_i)$. Введём модифицированный таргет $\hat{y}_i = \log(y_i + 1)$ и модель $\hat{a}(x_i)$, которая обучается на $\hat{y}_i$ и предсказывает логарифм таргета. Тогда исходное предсказание восстанавливается как $a(x_i) = \exp(\hat{a}(x_i)) - 1$, то есть $\hat{a}(x_i) = \log(a(x_i) + 1)$.

Запишем RMSLE для модели $a$:

$$\text{RMSLE}(X, y, a) = \sqrt{\frac{1}{\ell}\sum_{i=1}^{\ell} \big(\log(y_i + 1) - \log(a(x_i) + 1)\big)^2}$$

Подставим выражения через $\hat{y}_i$ и $\hat{a}(x_i)$:

$$\text{RMSLE}(X, y, a) = \sqrt{\frac{1}{\ell}\sum_{i=1}^{\ell} \big(\hat{y}_i - \hat{a}(x_i)\big)^2} = \sqrt{\text{MSE}(X, \hat{y}, \hat{a})}$$

Корень — монотонно возрастающая функция, поэтому минимизация RMSLE по $\hat{a}$ эквивалентна минимизации MSE по $\hat{a}$. Значит, обучая линейную регрессию (которая минимизирует MSE) на логарифме таргета, мы тем самым минимизируем RMSLE на исходном таргете. $\blacksquare$


**(0.1 балл)** Итак, мы смогли свести задачу оптимизации RMSLE к задаче оптимизации MSE, которую мы умеем решать! Кроме того, у логарифмирования таргета есть еще одно полезное свойство. Чтобы его увидеть, добавьте к нашей выборке колонку `log_trip_duration` (воспользуйтесь `np.log1p`) и нарисуйте гистограмму модифицированного таргета по обучающей выборке. Удалите колонку со старым таргетом.

```python
#╰( ͡° ͜ʖ ͡° )つ──☆*:・ﾟ
df['log_trip_duration'] = np.log1p(df['trip_duration'])
df = df.drop('trip_duration', axis=1)

plt.figure(figsize=(10, 5))
plt.hist(df['log_trip_duration'], bins=100)
plt.title("Распределение логарифма длительности поездок")
plt.xlabel("log(1 + длительность поездки)")
plt.ylabel("Количество поездок")
plt.tight_layout()
plt.show()
```

**(0.1 балл)** Чтобы иметь некоторую точку отсчета, давайте посчитаем значение метрики MSE при наилучшем константном предсказании:

```python
#╰( ͡° ͜ʖ ͡° )つ──☆*:・ﾟ
const_pred = df['log_trip_duration'].mean()
mse_const = ((df['log_trip_duration'] - const_pred) ** 2).mean()
print("MSE при константном предсказании:", mse_const)
```

```python
def show_circles_on_map(data, latitude_column, longitude_column, color):
    """
    The function draws map with circles on it.
    The center of the map is the mean of coordinates passed in data.

    data: DataFrame that contains columns latitude_column and longitude_column
    latitude_column: string, the name of column for latitude coordinates
    longitude_column: string, the name of column for longitude coordinates
    color: string, the color of circles to be drawn
    """

    location = (data[latitude_column].mean(), data[longitude_column].mean())
    m = folium.Map(
        location=location,
        tiles='https://basemaps.cartocdn.com/rastertiles/voyager/{z}/{x}/{y}.png?key=cb1_49uj_1_a3d38ef712858e5c05155118',
        attr='© OpenStreetMap contributors © CARTO'
        )
    # Я модифицировал генератор карт, чтобы он показывал картинку. простите, пожалуйста!

    for _, row in data.iterrows():
        folium.Circle(
            radius=100,
            location=(row[latitude_column], row[longitude_column]),
            color=color,
            fill_color=color,
            fill=True
        ).add_to(m)

    return m
```

**(0.1 балла)** Как мы все прекрасно помним, $t = s / v_{\text{ср}}$, поэтому очевидно, что самым сильным признаком будет расстояние, которое необходимо проехать. Мы не можем посчитать точное расстояние, которое необходимо преодолеть такси, но мы можем его оценить, посчитав кратчайшее расстояние между точками начала и конца поездки. Чтобы корректно посчитать расстояние между двумя точками на Земле, можно использовать функцию `haversine`. Посчитайте кратчайшее расстояние для объектов и запишите его в колонку `haversine`:

```python
#╰( ͡° ͜ʖ ͡° )つ──☆*:・ﾟ
def haversine(lat1, lon1, lat2, lon2):
    R = 6371
    lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    a = np.sin(dlat / 2) ** 2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon / 2) ** 2
    return R * 2 * np.arcsin(np.sqrt(a))

for d in [train_df, test_df]:
    d['haversine'] = haversine(
        d['pickup_latitude'].values,
        d['pickup_longitude'].values,
        d['dropoff_latitude'].values,
        d['dropoff_longitude'].values
    )
```

**(0.1 балла)** Так как мы предсказываем логарифм времени поездки и хотим, чтобы наши признаки были линейно зависимы с этой целевой переменной, нам нужно логарифмировать расстояние: $\log t = \log s - \log{v_{\text{ср}}}$. Запишите логарифм `haversine` в отдельную колонку:

```python
#╰( ͡° ͜ʖ ͡° )つ──☆*:・ﾟ
for d in [train_df, test_df]:
    d['log_haversine'] = np.log1p(d['haversine'])
```

Напишите трансформер, который сначала разбивает показанную на рисунке область на ячейки, а затем создает два признака: номер ячейки, в которой началась поездка, и номер ячейки, в которой закончилась поездка. Количество строк и столбцов выберите самостоятельно.

Обратите внимание, что все вычисления должны быть векторизованными, трансформер не должен модифицировать передаваемую ему выборку inplace, а все необходимые статистики (если они вдруг нужны) нужно считать только по обучающей выборке в методе `fit`:

```python
from sklearn.base import BaseEstimator, TransformerMixin

class MapGridTransformer(BaseEstimator, TransformerMixin):
    def __init__(self, lat_min=40.7, lat_max=40.82, lon_min=-74.02, lon_max=-73.90,
                 n_rows=20, n_cols=20):
        self.lat_min = lat_min
        self.lat_max = lat_max
        self.lon_min = lon_min
        self.lon_max = lon_max
        self.n_rows = n_rows
        self.n_cols = n_cols

    def fit(self, X, y=None):
        return self

    def transform(self, X):
        X = X.copy()   # не модифицируем входной DataFrame

        lat_step = (self.lat_max - self.lat_min) / self.n_rows
        lon_step = (self.lon_max - self.lon_min) / self.n_cols

        for prefix in ['pickup', 'dropoff']:
            lat = X[f'{prefix}_latitude'].values
            lon = X[f'{prefix}_longitude'].values

            # номер строки и столбца ячейки
            row = ((lat - self.lat_min) / lat_step).astype(int)
            col = ((lon - self.lon_min) / lon_step).astype(int)

            # маска: точка попала в сетку
            in_grid = (
                (row >= 0) & (row < self.n_rows) &
                (col >= 0) & (col < self.n_cols)
            )

            # номер ячейки: row * n_cols + col, вне сетки — -1
            cell = np.where(in_grid, row * self.n_cols + col, -1).astype(int)

            X[f'{prefix}_cell'] = cell

        return X

grid = MapGridTransformer(n_rows=20, n_cols=20)
grid.fit(train_df)

train_df = grid.transform(train_df)
test_df = grid.transform(test_df)
```

**Вопрос (0.5 балла)**: **(?)** Почему такой признак потенциально полезный? Почему линейная модель не может самостоятельно "вытащить" эту информацию, ведь у нее в распоряжении есть признаки "из какой ячейки началась поездка" и "в какой ячейке закончилась поездка"?

**Ответ**:
Признак `route_id` кодирует **маршрут целиком** — конкретную пару «откуда → куда», а не старт и финиш по отдельности. Это важно, потому что длительность поездки зависит не только от того, где ты сел и где вышел, но и от того, **что между этими точками**: едешь ли ты через мост, объезжаешь ли парк, попадаешь ли на конкретный перекрёсток с вечной пробкой.

Линейная модель складывает вклады признаков **независимо**:
`w_A * [старт в A] + w_B * [финиш в B]`.
Она физически не может выучить, что именно маршрут A→B долгий, а A→C при тех же A и B — короткий. Для модели «старт в A» — это одна и та же величина, куда бы ты ни ехал.

Interaction-признак даёт каждой паре **свой отдельный коэффициент**. Модель получает возможность сказать: «поездка из ячейки 5 в ячейку 12 — это +0.3 к логарифму времени», отдельно от «старт в ячейке 5» и «финиш в ячейке 12». Без interaction такое возможно только для нелинейных моделей (деревья, kNN), а у нас линейная — поэтому interaction критичен.

На практике прирост MSE небольшой (0.2192 → 0.2166), потому что 88% поездок попадают в -1 (нетипичные пары), и только 12% получают осмысленный `route_id`. Но даже на этой доле признак даёт сигнал, которого у модели раньше не было.

**Вопрос (0.2 балла)**: **(?)** Найдите, что такое манхэттенское расстояние и почему оно так называется. Как оно нам может помочь?

**Ответ**:
Манхэттенское расстояние (city block distance, L1-норма) — это сумма абсолютных разностей координат:

d(A, B) = |x_A - x_B| + |y_A - y_B|

Название происходит от района Манхэттен в Нью-Йорке, где улицы образуют регулярную прямоугольную сетку (grid). Пешеход или машина не может двигаться по прямой — путь идёт вдоль улиц, то есть по горизонтали и вертикали. Поэтому реальное расстояние между двумя точками — это сумма «пройденного по авеню» и «пройденного по стрит».

Для нашей задачи это полезно, потому что такси в Нью-Йорке тоже ездит по улицам, а не по прямой. Haversine считает расстояние по прямой (сквозь здания и реку), а манхэттенское — приближённо повторяет реальный маршрут. Значит, оно должно лучше коррелировать с временем поездки.

```python
#╰( ͡° ͜ʖ ͡° )つ──☆*:・ﾟ
# Центр Манхэттена и угол поворота улиц
LAT0, LON0 = 40.7580, -73.9855
THETA = np.radians(29)
COS_LAT = np.cos(np.radians(LAT0))   # ~0.757

def rotate_coords(lat, lon):
    lon_km = (lon - LON0) * COS_LAT    # приводим к одному масштабу с широтой
    lat_km = (lat - LAT0)
    x =  lon_km * np.cos(THETA) + lat_km * np.sin(THETA)
    y = -lon_km * np.sin(THETA) + lat_km * np.cos(THETA)
    return x, y

def manhattan_distance(lat1, lon1, lat2, lon2):
    x1, y1 = rotate_coords(lat1, lon1)
    x2, y2 = rotate_coords(lat2, lon2)
    return np.abs(x1 - x2) + np.abs(y1 - y2)

for d in [train_df, test_df]:
    d['manhattan'] = manhattan_distance(
        d['pickup_latitude'].values, d['pickup_longitude'].values,
        d['dropoff_latitude'].values, d['dropoff_longitude'].values
    )
    d['log_manhattan'] = np.log1p(d['manhattan'])

corr_hav = train_df['log_haversine'].corr(train_df['log_trip_duration'])
corr_man = train_df['log_manhattan'].corr(train_df['log_trip_duration'])
print(f"Corr log_haversine: {corr_hav:.4f}")
print(f"Corr log_manhattan: {corr_man:.4f}")
```

Важно: производительность библиотеки существенно зависит от количества шестиугольников на карте (определяется параметром resolution). Подберите такое разрешение, при котором ваш код будет работать за приемлемое время.

```python
import h3
from sklearn.base import BaseEstimator, TransformerMixin

class H3Transformer(BaseEstimator, TransformerMixin):
    def __init__(self, resolution=9):
        self.resolution = resolution

    def fit(self, X, y=None):
        # Считаем частоты ячеек ТОЛЬКО на train (если нужно)
        return self

    def transform(self, X):
        X = X.copy()
        res = self.resolution

        # 1. H3-ячейки старта и финиша
        X['pickup_h3'] = [h3.latlng_to_cell(lat, lon, res)
                          for lat, lon in zip(X['pickup_latitude'], X['pickup_longitude'])]
        X['dropoff_h3'] = [h3.latlng_to_cell(lat, lon, res)
                           for lat, lon in zip(X['dropoff_latitude'], X['dropoff_longitude'])]

        # 2. Расстояние в шагах сетки (grid distance)
        X['h3_distance'] = [h3.grid_distance(p, d) if p != d else 0
                            for p, d in zip(X['pickup_h3'], X['dropoff_h3'])]

        # 3. (опционально) Число поездок в ячейке старта — считать ТОЛЬКО на train!
        # Это делается в fit(), чтобы не было утечки.

        return X
```
